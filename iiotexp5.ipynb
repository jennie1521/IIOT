```python

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("farm_temperature_humidity_3_months.csv")

# Select Farm 1
farm = df[df["Farm"] == "Farm 1"].copy()
farm["Timestamp"] = pd.to_datetime(farm["Timestamp"])
farm = farm.sort_values("Timestamp")

def plot_sensor(column, hours, unit):
    data = farm.head(hours)

    plt.figure(figsize=(10, 4))
    plt.plot(data["Timestamp"], data[column], marker="o")
    plt.title(f"{column} - {hours} Hours")
    plt.xlabel("Time")
    plt.ylabel(unit)
    plt.xticks(rotation=45)
    plt.grid(True)
    plt.tight_layout()
    plt.show()

# Graph 1: Temperature - 24 hours
plot_sensor("Temperature (°C)", 24, "°C")

# Graph 2: Temperature - 48 hours
plot_sensor("Temperature (°C)", 48, "°C")

# Graph 3: Humidity - 24 hours
plot_sensor("Humidity (%)", 24, "%")

# Graph 4: Humidity - 48 hours
plot_sensor("Humidity (%)", 48, "%")
```
